```rust
// Step 1: Create a mutable String
let mut pipeline_stage = String::from("Ingestion");

// Step 2: Borrow a mutable reference (&mut) to modify the underlying data
let stage_ref = &mut pipeline_stage;

// Step 3: Mutate the data by pushing additional text
stage_ref.push_str(" -> Transformation");

// Step 4: Inspect the modified result
println!("Updated Pipeline: {}", stage_ref);
```

```rust
{
    // Step 1: Create a mutable String
    let mut pipeline_stage = String::from("Ingestion");

    // Step 2: Borrow a mutable reference (&mut) to modify the underlying data
    let stage_ref = &mut pipeline_stage;

    // Step 3: Mutate the data by pushing additional text
    stage_ref.push_str(" -> Transformation");

    // Step 4: Inspect the modified result
    println!("Updated Pipeline: {}", stage_ref);
}
```

```rust
{
    // Step 1: Initialize a mutable dataset identifier
    let mut dataset_name = String::from("users_data");

    // Step 2: Create the first mutable reference
    let ref1 = &mut dataset_name;

    // Step 3: Attempt to create a second mutable reference while ref1 is still active
    let ref2 = &mut dataset_name;

    // Step 4: Use both references to trigger the conflict
    println!("Reference 1: {}, Reference 2: {}", ref1, ref2);
}
```

```rust
{
    // Step 1: Initialize the mutable dataset
    let mut dataset_name = String::from("users_data");

    // Step 2: First borrow scope - modify the name
    {
        let ref1 = &mut dataset_name;
        ref1.push_str("_v1");
        println!("Modified via ref1: {}", ref1);
    } // ref1 goes out of scope here and releases its borrow!

    // Step 3: Second borrow scope - safe to borrow again
    let ref2 = &mut dataset_name;
    ref2.push_str("_cleaned");
    println!("Modified via ref2: {}", ref2);
}
```

```rust
{
    // Step 1: Create original data
    let original_metrics = String::from("Accuracy: 95%");

    // Step 2: Perform a deep copy using .clone()
    let mut cloned_metrics = original_metrics.clone();

    // Step 3: Modify only the cloned copy
    cloned_metrics.push_str(" | Precision: 92%");

    // Step 4: Verify both coexist independently without ownership conflicts
    println!("Original: {}", original_metrics);
    println!("Cloned & Updated: {}", cloned_metrics);
}
```
